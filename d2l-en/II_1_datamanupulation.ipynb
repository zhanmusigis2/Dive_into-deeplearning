```python
before=id(x)
x=x+y
id(x)==before# this is an out-place operation, first, we create a new tensor x+y， then Y is assigned to the new space
```

```python
before=id(x)
x+=y
id(x)==before #this is an in-place operation, and the a will not point to a new space
```

```python
Z=torch.zeros_like(y)
before=id(Z)
Z[:]=x+y
id(Z)==before# this is an in-place operation and the space to which pointed by Z is not changed
```

```python
Z=torch.zeros_like(y)
before=id(Z)
Z=x+y
id(Z)==before# this is an out-place operation and the space to which pointed by Z is changed
```
